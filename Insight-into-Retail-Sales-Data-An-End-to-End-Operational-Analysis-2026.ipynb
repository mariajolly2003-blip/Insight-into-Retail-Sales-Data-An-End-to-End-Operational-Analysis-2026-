### 6. HANDLING OULIERS WITH IQR METHOD ###

Boxplots were used to visually examine the distribution of the numerical variables and identify potential outliers. The boxplot displays the median, first quartile (Q1), third quartile (Q3), and the interquartile range (IQR).

Observations falling below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are displayed as individual points and are considered potential outliers.

The boxplots were used as a preliminary diagnostic rather than as an automatic basis for removing observations.Therefore, potential outliers were reviewed in the context of the dataset and were not removed solely because they appeared outside the boxplot whiskers.

```python
numeric_cols = [
     "age",
    "quantity",
    "unit_price",
    "discount_pct",
    "sales_amount",
    "profit",
    "customer_satisfaction",
    "shipping_cost",
    "days_to_ship"
]
outlier_summary = pd.DataFrame(index=numeric_cols)
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outlier_summary.loc[col, "Q1"] = Q1
    outlier_summary.loc[col, "Q3"] = Q3
    outlier_summary.loc[col, "Lower_Bound"] = lower
    outlier_summary.loc[col, "Upper_Bound"] = upper
    outlier_summary.loc[col, "Outlier_Count"] = (
        (df[col] < lower) | (df[col] > upper)
    ).sum()

outlier_summary
```

This function handles outliers using the IQR (Interquartile Range) method. First, it calculates Q1 (25th percentile) and Q3 (75th percentile), then finds the IQR as Q3 - Q1. The lower bound is calculated as Q1 - 1.5 × IQR and the upper bound as Q3 + 1.5 × IQR. Any values outside these limits are considered outliers. Instead of deleting them, the clip() function replaces values below the lower bound with the lower bound and values above the upper bound with the upper bound. This reduces the effect of extreme values while keeping all the rows in the dataset.

Using the Interquartile Range method it calculates the statistical boundaries for every numerical feature in our dataset.   
First, our transactional financial dimensions: Notice that unit_price, sales_amount, and profit have exceptionally high outlier counts, all hovering around 500 records each. The upper limit for a standard single order sale is capped mathematically at roughly $122,600. Transactions exceeding this are not input errors; they reflect our corporate B2B clients placing massive, high-volume orders.

Second, our operational metrics: Columns like customer age, shipping_cost, and days_to_ship display remarkably low outlier frequencies, indicating a stable baseline for our standard logistics data.

## 8.4 Multivariate Analysis ##

#### 8.4.1. Average Shipping Time: Returned vs. Kept Orders by Category  ####

```python
grouped_data = df.groupby('product_category').agg(
    avg_shipping_days=('days_to_ship', 'mean'),
    return_rate_ratio=('return_flag', 'mean')
).sort_values(by='avg_shipping_days') # Sort by shipping speed to make trends visible

fig, ax1 = plt.subplots(figsize=(10, 5))

# Plot the Average Shipping Days
sns.barplot(
    x=grouped_data.index, 
    y=grouped_data['avg_shipping_days'], 
    ax=ax1, 
    color='blue', 
    alpha=0.75
)
ax1.set_title('Supply Chain Impact: Shipping Speed vs. Customer Return Rates')
ax1.set_xlabel('Product Category')
ax1.set_ylabel('Average Days to Ship (Bars)', color='blue')
ax1.tick_params(axis='y', labelcolor='crimson')

ax2 = ax1.twinx()

# Plot the  Return Rate Ratio
sns.lineplot(
    x=grouped_data.index, 
    y=grouped_data['return_rate_ratio'], 
    ax=ax2, 
    color='crimson', 
    marker='o', 
    linewidth=2.5, 
    markersize=8
)

ax2.set_ylabel('Return Rate Ratio (Line: 0.0 - 1.0)', color='crimson')
ax2.tick_params(axis='y', labelcolor='crimson')

plt.tight_layout()
plt.show()
```

"To analyze  supply chain's direct financial impact on the company, builted a dual-axis combo chart. The blue bars display the average number of days it takes our team to ship out an item, sorted from fastest to slowest. The red line overlay tracks the customer return rate for those exact same categories.This visualization reveals a highly targeted operational bottleneck. In categories where immediate fulfillment is critical—specifically Electronics and Groceries—any increase in shipping time creates an immediate, severe penalty spike in customer returns. Our slowest category, Electronics, triggers our highest return rate at 16%.However, we also see that categories like Furniture and Beauty are far more forgiving of slight delays, maintaining low return rates around 13.5%. Therefore, our recommendation to management is clear: to maximize our net profitability and cut down reverse-logistics losses, we must prioritize optimization resources specifically to speed up Electronics and Groceries fulfillment times
